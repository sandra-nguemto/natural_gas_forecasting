# Data Cleaning Documentation
Raw data was obtained from: [eia.gov](https://www.eia.gov/dnav/ng/ng_cons_sum_a_EPG0_vgt_mmcf_m.htm) and contains monthly U.S. natural gas consumption data (for residential, commercial, industrial, vehice fuel and electric power consumers) by state from January 2001 to August 2025. The dataset required cleaning to ensure proper formatting and data types for analysis.

```python
import os
os.getcwd()
```

```python
import pandas as pd

df_raw = pd.read_csv("../data/raw/consumer_gas_monthly.csv", header=None)
df_raw.head()
```

```python
#Extracting the header
header = df_raw.iloc[2].tolist()
header
```

```python
# Dropping the first three rows and assigning the header    
df = df_raw[3:].copy()
df.columns = header
df.head()
```

```python
# Converting the 'Date' column to datetime format
df["Date"] = pd.to_datetime(df["Date"], format="%b-%Y")
df.head()
```

```python
# Simplify the long column names
def clean_col(col):
    if col == "Date":
        return "Date"
    if "U.S." in col:
        return "US_Total"
    # everything between "in " and " (Including"
    return col.split("in ")[1].split(" (Including")[0].strip()

df.columns = [clean_col(col) for col in df.columns]
df.head()
```

```python
# Converting all numeric columns to float
for col in df.columns:
    if col != "Date":
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
#Set date as index and sort
df = df.set_index("Date").sort_index()
```

```python
df.head()
```

```python
#Saved the cleaned dataset
df.to_csv("../data/processed/consumer_gas_monthly_cleaned.csv")
```
